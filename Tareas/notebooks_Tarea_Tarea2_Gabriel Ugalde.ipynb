# Estrellas: Luminosidad y distancia de las estrellas

```python
# ============================================================
# Nombre completo: Gabriel Enrique Ugalde Agüero
# Rut: 22.601.194-3
# ============================================================
import numpy as np

# Semilla (elige un número entre 0 y 100000)
su_numero = 12345
rng = np.random.default_rng(seed=su_numero)

# Datos del problema
L = [1.0, 2.5, 0.8, 5.0, 10.0] # Luminosidades (Lsol)
d = [10, 25, 15, 40, 60] # Distancias (parsec)

# Aqui empieza el codigo
L = np.array(L)
d = np.array(d)

print("Arreglo de luminosidades:", L)
print("Arreglo de distancias:", d)

def calcular_flujo(L, d):
    """
    Calcula el flujo observado de una estrella.
    
    Parámetros:
        L : array o número - Luminosidad de la estrella
        d : array o número - Distancia a la estrella
    
    Retorna:
        F : flujo en W/m²
    """
    return L / (4 * np.pi * d**2)

# Calculamos el flujo de todas las estrellas
F = calcular_flujo(L, d)
print("\nFlujos observados (W/m²):", F)

indice_max = np.argmax(F)
print("\nÍndice de la estrella con mayor flujo:", indice_max)

indice_min = np.argmin(F)
print("Índice de la estrella con menor flujo:", indice_min)

flujo_promedio = np.mean(F)
flujo_mediana = np.median(F)
print("\nFlujo promedio:", flujo_promedio)
print("Mediana del flujo:", flujo_mediana)

limite = 1e-11
mascara = F > limite # True = se puede detectar
print("\nMáscara de detección:", mascara)

indices_detectadas = np.where(mascara)[0]
print("Índices de estrellas detectadas:", indices_detectadas)

orden = np.argsort(F)[::-1] # [::-1] invierte el orden
print("\nÍndices ordenados (mayor → menor flujo):", orden)
print("Flujos ordenados:", F[orden])
```

# Transito planetario

```python
# ============================================================
# Nombre completo: Gabriel Enrique Ugalde Agüero
# Rut: 22.601.194-3
# ============================================================
import numpy as np

# Semilla (elige un número entre 0 y 100000)
su_numero = 12345
rng = np.random.default_rng(seed=su_numero)

# Generar el arreglo de tiempo (10 años, 500 puntos)
tiempo = np.linspace(0, 10, 500) # 500 mediciones equidistantes

# Flujo inicial = 1 (normalizado)
flujo = np.ones_like(tiempo)

# Agregar ruido gaussiano (desviación estándar = 0.002)
ruido = rng.normal(0, 0.002, size=tiempo.shape)
flujo = flujo + ruido

# Simular el tránsito entre t = 4.5 y t = 5.0 años
# (disminuir el flujo en 0.02)
mascara_transito = (tiempo >= 4.5) & (tiempo <= 5.0)
flujo[mascara_transito] = flujo[mascara_transito] - 0.02

# Aqui inicia el codigo
flujo_promedio = np.mean(flujo)
print("Flujo promedio:", flujo_promedio)

sigma = np.std(flujo)
print("Desviación estándar:", sigma)

flujo_min = np.min(flujo)
indice_min = np.argmin(flujo)
print("\nValor mínimo del flujo:", flujo_min)
print("Índice del mínimo:", indice_min)

tiempo_min = tiempo[indice_min]
print("Tiempo del mínimo:", tiempo_min)

mascara_3sigma = flujo < (flujo_promedio - 3 * sigma)
puntos_3sigma = flujo[mascara_3sigma]
tiempos_3sigma = tiempo[mascara_3sigma]

print("\nPuntos que cumplen F < promedio - 3σ:")
print("Cantidad de puntos:", len(puntos_3sigma))
print("Tiempos de esos puntos:", tiempos_3sigma)

print("Número de puntos identificados:", len(puntos_3sigma))

# Usamos el valor mínimo como aproximación de F_transit
F_transit = flujo_min
Delta_F = 1 - F_transit
print("\nProfundidad del tránsito (ΔF):", Delta_F)

razon_radios = np.sqrt(Delta_F)
print("Razón de radios Rp/R*:", razon_radios)

print("\n--- Comparación ---")
print("Intervalo real del tránsito: 4.5 a 5.0 años")
print("Tiempos detectados con 3σ:", tiempos_3sigma)
```
